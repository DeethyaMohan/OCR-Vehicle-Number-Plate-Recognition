```python
from ultralytics import YOLO
import cv2
import pytesseract
```

```python
import matplotlib.pyplot as plt
import re
```

```python
model = YOLO('best.pt')
# pytesseract.pytesseract.tesseract_cmd = r'D:\Program Files\tesseract.exe'
```

```python
result = model('fancy_number_plate_bfbc501f34.jpg')

# use a different variable name to avoid shadowing the `re` module
for res in result:
    res.save(filename='result.jpg')

    for box in res.boxes.xyxy.tolist():
        x1, y1, x2, y2 = box  # [x1, y1, x2, y2]

# convert coordinates to integers once after the loop
x1, y1, x2, y2 = map(int, (x1, y1, x2, y2))

img = cv2.imread('fancy_number_plate_bfbc501f34.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img = cv2.GaussianBlur(img,(5,5),0)
img = cv2.threshold(img,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)[1]

img = img[y1:y2, x1:x2]



data = pytesseract.image_to_data(img, output_type=pytesseract.Output.DICT)
for i, text in enumerate(data['text']):
    
    if data['conf'][i] > 50 and re.search(r'[A-Z]{2}\d{2}[A-Z]{2}\d{4}', text):
        print(text)
plt.imshow(img)



```
